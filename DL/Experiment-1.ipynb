```python

import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Flatten, Dense

corpus = ["I love deep learning","Deep learning models can learn word embeddings",
"Word embeddings capture semantic meaning of words","Neural networks are powerful for NLP tasks"]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)
word_index = tokenizer.word_index
sequences = tokenizer.texts_to_sequences(corpus)
X = pad_sequences(sequences,padding='post')

print("Vocabulary:",word_index)
print("Sequences:",sequences)
print("Padded Sequences:\n",X)

model = Sequential([Embedding(len(word_index)+1,8),
                    Flatten(),
                    Dense(10,activation='relu'),
                    Dense(1,activation='sigmoid')])

model.compile(optimizer='adam',loss='binary_crossentropy')

X = np.array(X)
y = np.array([1, 0, 1, 0])

model.fit(X, y, epochs=20, verbose=0)


embeddings = model.layers[0].get_weights()[0]
for word in ["deep","learning"]:
    print(f"Embedding for word '{word}':\n",embeddings[word_index[word]])
```
