```python
from tensorflow.keras.preprocessing.text import *
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import *
import numpy as np

data = ["I love deep learning", 
        "Deep learning models",
        "Word embeddings capture meaning", 
        "Neural networks for NLP"]

t = Tokenizer()
t.fit_on_texts(data)
X = pad_sequences(t.texts_to_sequences(data))

print("Vocabulary:", t.word_index)
print("Padded:", X)

m = Sequential([Embedding(len(t.word_index)+1, 8), 
                Flatten(),
                Dense(10, activation='relu'), 
                Dense(1, activation='sigmoid')])

m.compile(optimizer='adam', loss='binary_crossentropy')
y=np.array([1,0,1,0])
m.fit(X,y, epochs=20)

e = m.layers[0].get_weights()[0]
print("Embedding for Deep:",e[t.word_index['deep']])
print("\nEmbedding for Learning:" ,e[t.word_index['learning']])
```
