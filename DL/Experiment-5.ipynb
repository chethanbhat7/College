```python

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Flatten, Dense

documents = ["I love deep learning","Deep learning is amazing",
"I enjoy machine learning","Natural language processing is fun",
"I dislike bugs in code","Debugging code can be hard",
"I hate syntax errors","Errors make me frustrated"]
labels = [1,1,1,1,0,0,0,0]

tokenizer = Tokenizer(num_words=50)
tokenizer.fit_on_texts(documents)
sequences = tokenizer.texts_to_sequences(documents)
X = tf.convert_to_tensor(pad_sequences(sequences,maxlen=6))

print("Word Index:",tokenizer.word_index)
print("Sequences:",sequences)
print("Padded Sequences:\n",X.numpy())

model = Sequential([Embedding(50,8),Flatten(),
                    Dense(10,activation='relu'),
                    Dense(1,activation='sigmoid')])
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
model.fit(X,tf.constant(labels),epochs=20)


test_sequence = tokenizer.texts_to_sequences(["I love coding"])
test_padded = tf.convert_to_tensor(pad_sequences(test_sequence,maxlen=6))
prediction = model.predict(test_padded)[0][0]

print("Prediction:",prediction)
print("Class:","Positive" if prediction>0.5 else "Negative")
```
