```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import *
import numpy as np

documents = ["I love deep learning","Deep learning is amazing",
"I enjoy machine learning","Natural language processing is fun",
"I dislike bugs in code","Debugging code can be hard",
"I hate syntax errors","Errors make me frustrated"]
y = np.array([1,1,1,1,0,0,0,0])

tokenizer = Tokenizer()
tokenizer.fit_on_texts(documents)
X = pad_sequences(tokenizer.texts_to_sequences(documents),maxlen=6)

print(tokenizer.word_index, X)

model = Sequential([Embedding(len(tokenizer.word_index)+1,8),
                    Flatten(),
                    Dense(10,'relu'),
                    Dense(1,'sigmoid')])

model.compile(optimizer='adam',loss='binary_crossentropy')
model.fit(X,y,epochs=20)

test = pad_sequences(tokenizer.texts_to_sequences(["I love coding"]),maxlen=6)

prediction = model.predict(test)[0][0]
print("Prediction:",prediction)
print("Class:","Positive" if prediction>0.5 else "Negative")
```
