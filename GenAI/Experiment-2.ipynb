```python
import gensim.downloader as api
import numpy as np
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from plotly.subplots import make_subplots
import plotly.graph_objects as go

model = api.load("glove-wiki-gigaword-50")

words = ["queen","throne","prince","daughter","elizabeth",
         "princess","kingdom","monarch"]

X = np.array([model[w] for w in words])

pca = PCA(2).fit_transform(X)
tsne = TSNE(2, perplexity=5, random_state=42).fit_transform(X)
```

```python
go.Figure(go.Scatter(
    x=pca[:,0], y=pca[:,1],
    mode="markers+text",
    text=words
)).update_layout(title="PCA",xaxis_title='PCA1',yaxis_title="PCA2").show()
go.Figure(go.Scatter(
    x=tsne[:,0], y=tsne[:,1],
    mode="markers+text",
    text=words
)).update_layout(title='TSNE',xaxis_title='Dim1',yaxis_title='Dim2').show()
```
