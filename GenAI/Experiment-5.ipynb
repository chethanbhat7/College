```python
import gensim.downloader as api

# Load model
model = api.load("glove-wiki-gigaword-50")

# Input seed word
seed = input("Enter a word: ")

# Get similar words
similar = [w[0] for w in model.most_similar(seed, topn=5)]

# Create paragraph
paragraph = f"""
The {seed} was known for its connection with {similar[0]} and {similar[1]}.
People often related it to {similar[2]} and {similar[3]}.
In many stories, {similar[4]} played an important role alongside the {seed}.
"""

print("\nGenerated Paragraph:\n")
print(paragraph)
```
