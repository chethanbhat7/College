```python
from pydantic import BaseModel
import json

class InstitutionDetails(BaseModel):
    founder: str | None
    founded: str | None
    branches: list[str] | None
    number_of_employees: int | None
    summary: str | None

with open("institution.json") as f:
    data = json.load(f)

name = input("Enter institution name: ").strip()

details = InstitutionDetails(**data[name])
print(f"Founder: {details.founder}")
print(f"Founded: {details.founded}")
print(f"Branches: {', '.join(details.branches)}")
print(f"Number of Employees: {details.number_of_employees}")
print(f"Summary: {details.summary}")
```
