```python
import json
data = json.load(open("institutions.json"))
name = input("Enter institution name: ").strip()
if name in data:
 d = data[name]
 print(
 f"Founder: {d.get('founder', 'N/A')}\n"
 f"Founded: {d.get('founded', 'N/A')}\n"
 f"Branches: {', '.join(d.get('branches', [])) or 'N/A'}\n"
 f"Employees: {d.get('number_of_employees', 'N/A')}\n"
 f"Summary:\n{d.get('summary', 'N/A')}"
 )
elif name:
 print(f"'{name}' not found in local dataset.")
else:
 print("Please enter a valid institution name.")
```
