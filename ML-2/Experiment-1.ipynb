```python
import pandas as pd

d=pd.read_csv("enjoysport.csv")
X=d.iloc[:,:-1].values
Y=d.iloc[:,-1].values

h=list(X[0])
for i in range(len(X)):
    if Y[i]=="yes":
        for j in range(len(h)):
            if h[j]!=X[i][j]: h[j]="?"

print("Final Hypothesis:",h)
```

```python
import pandas as pd

d=pd.read_csv("enjoysport.csv")
X=d.iloc[:,:-1].values
Y=d.iloc[:,-1].values

S=list(X[0])
G=[['?']*len(S) for i in S]

for i in range(len(X)):
    for j in range(len(S)):
        if Y[i]=="yes" and S[j]!=X[i][j]:
            S[j]='?'
            G[j][j]='?'
        elif Y[i]=="no" and S[j]!=X[i][j]:
            G[j][j]=S[j]

print("S =",S)
print("G =")
for g in G:
    if g!=['?']*len(S): print(g)
```
