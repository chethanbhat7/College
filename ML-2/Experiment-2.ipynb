```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import BaggingClassifier,AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,classification_report

X,y=load_iris(return_X_y=True)
a,b,c,d=train_test_split(X,y,test_size=.3,random_state=42,stratify=y)
# Bagging
bag=BaggingClassifier(DecisionTreeClassifier(),n_estimators=10,random_state=42)
bag.fit(a,c)
p1=bag.predict(b)

# Boosting
boost=AdaBoostClassifier(DecisionTreeClassifier(max_depth=1),
                         n_estimators=10,random_state=42)
boost.fit(a,c)
p2=boost.predict(b)

print("Bagging Accuracy:",accuracy_score(d,p1))
print("Boosting Accuracy:",accuracy_score(d,p2))

print("\n--- Bagging Classification Report ---")
print(classification_report(d,p1,target_names=load_iris().target_names))

print("--- Boosting Classification Report ---")
print(classification_report(d,p2,target_names=load_iris().target_names))
```
