```python
import random

random.seed(0)
n=10
x=[random.random() for i in range(n)]
y=[i**2 for i in x]

I=sum(y)/n
exact=1/3

print("Sample\t x\t\t f(x)=x^2")
for i in range(n):
    print(i+1, "\t", f"{x[i]:.6f}\t", f"{y[i]:.6f}")

print("\nMonte Carlo estimate:",I)
print("Exact value:",exact)
print("Absolute error:",abs(I-exact))
```
