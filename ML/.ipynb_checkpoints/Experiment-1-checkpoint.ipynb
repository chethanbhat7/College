```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.DataFrame({
    'Age':[23,23,27,27,39,41,47,49,50,52,54,54,56,57,58,58,60,61],
    '%Fat':[7.8,9.5,17.8,25.9,26.5,27.2,27.4,28.8,30.2,31.2,31.4,32.9,33.4,34.1,34.6,35.7,41.2,42.5]
})

fat = df['%Fat']

print("Mean:", fat.mean())
print("Median:", fat.median())
print("Mode:", fat.mode()[0])
print("Std Dev:", fat.std())
print("Variance:", fat.var())
print("Range:", fat.max()-fat.min())

sns.histplot(fat, bins=10)
plt.show()
plt.boxplot(fat) 
plt.show()

q1, q3 = fat.quantile([.25, .75])
outliers = fat[(fat < q1-1.5*(q3-q1)) | (fat > q3+1.5*(q3-q1))]
print("Outliers:\n", outliers)

df['Age Group'] = pd.cut(df['Age'], [0,29,50,100], labels=['Young','Middle-aged','Older'])
counts = df['Age Group'].value_counts()

counts.plot(kind='bar')
plt.show()
```
