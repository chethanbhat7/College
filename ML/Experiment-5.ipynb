```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("tip.csv")
X, y = df["total_bill"].values, df["tip"].values

def lwr(xq, tau=10):
    w = np.exp(-(X - xq)**2 / (2 * tau**2))
    b = np.polyfit(X, y, 1, w=w)   # weighted linear fit
    return np.polyval(b, xq)

xr = np.linspace(min(X), max(X))
plt.scatter(X, y,alpha=0.5)
plt.plot(xr, [lwr(x) for x in xr])
plt.scatter(30, lwr(30))
plt.show()

print("Predicted Tip for $30:", round(lwr(30),2))
```
