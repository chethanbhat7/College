```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Load data
df = pd.read_csv("BostonHousing.csv")

# Select columns
X = df[['RM']]
y = df['MEDV']

# Train
model = LinearRegression().fit(X, y)

# Plot
plt.scatter(X, y, alpha=0.5, label='Actual')
plt.plot(X, model.predict(X), color='red', label='Trend')
plt.title("Linear Regression: RM vs MEDV")
plt.legend()

# Accuracy
print("R² Score:", model.score(X, y))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("auto-mpg.csv", na_values='?').dropna()

X, y = df['horsepower'], df['mpg']

model = np.poly1d(np.polyfit(X, y, 2))
xp = np.linspace(X.min(), X.max())

plt.scatter(X, y, alpha=0.5)
plt.plot(xp, model(xp), color='red')
plt.title("Polynomial Regression: Horsepower vs MPG")
plt.show()

print(f"Prediction for 100HP: {model(100):.2f}")
```
